```python
import duckdb, os, json, re
ROOT = r"C:\Users\ROG\Cyber_svvv_project"
db = duckdb.connect(os.path.join(ROOT, "data", "case.duckdb"), read_only=True)
results = []
def check(name, fn):
    try:
        ok, detail = fn(); results.append((("PASS" if ok else "FAIL"), name, detail))
    except Exception as e:
        results.append(("ERROR", name, str(e).splitlines()[0][:120]))
def one(sql): return db.sql(sql).fetchone()

# Structural groups (no labels)
db.sql("""
CREATE TEMP TABLE grp AS
WITH so AS (SELECT acct_id FROM features WHERE is_send_only),
a AS (SELECT dst acct_id FROM tx GROUP BY dst HAVING bool_and(src IN (SELECT acct_id FROM so))),
b AS (SELECT DISTINCT dst acct_id FROM tx WHERE src IN (SELECT acct_id FROM a)
        AND dst NOT IN (SELECT acct_id FROM features WHERE is_receive_only))
SELECT f.*, CASE WHEN f.is_send_only THEN 'send-only'
                 WHEN f.acct_id IN (SELECT acct_id FROM a) THEN 'A (L1-like)'
                 WHEN f.acct_id IN (SELECT acct_id FROM b) THEN 'B (L2-like)'
                 WHEN f.is_receive_only THEN 'receive-only' ELSE 'normal' END grp
FROM features f""")

# Fix 1: allocation no longer drops outflows
check("1. group B: no pass-through below 0.9",
      lambda: (lambda r: (r[0] == 0, f"below 0.9: {r[0]}, min={r[1]:.3f}"))(
          one("SELECT COUNT(*) FILTER (WHERE pass_through_share < 0.9), MIN(pass_through_share) FROM grp WHERE grp='B (L2-like)'")))
check("1. group A: pass-through still >= 0.9",
      lambda: (lambda r: (r[0] == 0, f"below 0.9: {r[0]}"))(
          one("SELECT COUNT(*) FILTER (WHERE pass_through_share < 0.9) FROM grp WHERE grp='A (L1-like)'")))
check("1. commission never above 1.0",
      lambda: (lambda r: (r[0] == 0, f"above 1.0: {r[0]}"))(
          one("SELECT COUNT(*) FROM features WHERE commission_ratio_median > 1.0")))

# Fix 2: sinks have no invented hold time
check("2. receive-only: median_hold_hours is NULL",
      lambda: (lambda r: (r[0] == r[1], f"{r[0]}/{r[1]} NULL"))(
          one("SELECT COUNT(*) FILTER (WHERE median_hold_hours IS NULL), COUNT(*) FROM grp WHERE grp='receive-only'")))

# Fix 3: device_consistency is yes/no (0 or 1), NULL when no outgoing tx
check("3. device_consistency only 0, 1 or NULL",
      lambda: (lambda r: (r[0] == 0, f"other values: {r[0]}"))(
          one("SELECT COUNT(*) FROM features WHERE device_consistency NOT IN (0, 1)")))
check("3. device_consistency NULL for receive-only",
      lambda: (lambda r: (r[0] == r[1], f"{r[0]}/{r[1]} NULL"))(
          one("SELECT COUNT(*) FILTER (WHERE device_consistency IS NULL), COUNT(*) FROM grp WHERE grp='receive-only'")))
check("3. mules with foreign IPs score 0",
      lambda: (lambda r: (r[0] == 0, f"mules scoring 1: {r[0]}"))(
          one("SELECT COUNT(*) FROM grp WHERE grp IN ('A (L1-like)','B (L2-like)') AND device_consistency = 1")))

# Fix 4: recurring senders counted by days (independent recomputation)
check("4. recurring_sender_share uses >= 3 distinct days",
      lambda: (lambda r: (r[0] == 0, f"mismatches: {r[0]} of {r[1]:,}"))(one("""
WITH by_cp AS (SELECT dst acct, src cp, COUNT(*) n_tx, COUNT(DISTINCT CAST(ts AS DATE)) n_days
               FROM tx GROUP BY 1, 2),
r AS (SELECT acct, SUM(CASE WHEN n_days >= 3 THEN n_tx ELSE 0 END)::DOUBLE / SUM(n_tx) by_days
      FROM by_cp GROUP BY acct)
SELECT COUNT(*) FILTER (WHERE abs(f.recurring_sender_share - r.by_days) > 1e-6), COUNT(*)
FROM features f JOIN r ON f.acct_id = r.acct""")))

# Fix 5: splits = distinct receivers (group A should be 3-6, group B exactly 1)
check("5. group A splits within 3-6",
      lambda: (lambda r: (r[0] == 0, f"outside 3-6: {r[0]}, median={r[1]}"))(
          one("SELECT COUNT(*) FILTER (WHERE split_count_median NOT BETWEEN 3 AND 6), MEDIAN(split_count_median) FROM grp WHERE grp='A (L1-like)'")))
check("5. group B splits = 1",
      lambda: (lambda r: (r[0] == 0, f"not 1: {r[0]}"))(
          one("SELECT COUNT(*) FILTER (WHERE split_count_median <> 1) FROM grp WHERE grp='B (L2-like)'")))

# Fixes 6-7: settings in the profile, gate recorded
prof = json.loads(one("SELECT definition::VARCHAR FROM scoring_profiles WHERE profile_id='v1-verified'")[0])
txt = json.dumps(prof).lower()
check("6. cut-offs present in the profile",
      lambda: (all(k in txt for k in ["odd", "recurring", "round", "victim"]),
               "odd-hour, recurring, round unit, victim-like keys found" if all(k in txt for k in ["odd","recurring","round","victim"]) else "some keys missing"))
check("7. reciprocity gate + Trust re-weighting in profile",
      lambda: ("reciprocity" in txt and "0.05" in txt, "gate found" if "reciprocity" in txt else "gate missing"))

# Fix 8: --db option exists
src = open(os.path.join(ROOT, "engine", "features.py"), encoding="utf-8").read()
check("8. features.py has a --db option", lambda: ("--db" in src, "found" if "--db" in src else "missing"))

print(db.sql("""
SELECT grp, COUNT(*) n,
  ROUND(MEDIAN(pass_through_share),3) pass_thru, ROUND(MEDIAN(commission_ratio_median),3) commission,
  MEDIAN(split_count_median) splits, ROUND(MEDIAN(median_hold_hours),1) hold_h,
  ROUND(AVG(device_consistency),3) device_cons, ROUND(MEDIAN(recurring_sender_share),3) recurring
FROM grp GROUP BY grp ORDER BY grp""").df().to_string(index=False))
db.close()

print()
for s, n, d in results: print(f"[{s:5}] {n:48} {d}")
print(f"\n{sum(s=='PASS' for s,_,_ in results)} PASS · {sum(s=='FAIL' for s,_,_ in results)} FAIL · {sum(s=='ERROR' for s,_,_ in results)} ERROR")
```
