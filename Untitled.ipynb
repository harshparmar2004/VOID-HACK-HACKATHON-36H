```python
import duckdb, glob, os

PROJECT = r"C:\Users\ROG\Cyber_svvv_project"
DB  = os.path.join(PROJECT, "data", "case.duckdb")
CSV = glob.glob(os.path.join(PROJECT, "data", "*.csv"))[0].replace("\\", "/")

db = duckdb.connect(DB, read_only=True)
results = []

def check(name, fn):
    try:
        ok, detail = fn()
        results.append((("PASS" if ok else "FAIL"), name, detail))
    except Exception as e:
        results.append(("ERROR", name, str(e).splitlines()[0][:120]))

def one(sql):
    return db.sql(sql).fetchone()

# ---------- 1. Tables ----------
tables = {r[0] for r in db.sql("SHOW TABLES").fetchall()}
expected = {"ingest_meta","rejects","accounts","tx","scoring_profiles","scores",
            "layer_links","rings","cases","case_outputs","bank_directory"}
leftovers = {"raw","l1_cand","l3_cand","t","acc"}
check("All expected tables exist", lambda: (expected <= tables, f"missing: {sorted(expected - tables) or 'none'}"))
check("No exploration leftovers",  lambda: (not (leftovers & tables), f"found: {sorted(leftovers & tables) or 'none'}"))
check("features not built yet (expected)", lambda: ("features" not in tables, "ok" if "features" not in tables else "features exists"))

# ---------- 2. tx ----------
check("tx: 2,000,000 rows, tx_key unique",
      lambda: (lambda r: (r == (2000000, 2000000), f"rows={r[0]:,} distinct tx_key={r[1]:,}"))(
          one("SELECT COUNT(*), COUNT(DISTINCT tx_key) FROM tx")))
check("tx: 4,502 duplicated-ID rows flagged",
      lambda: (lambda r: (r[0] == 4502, f"is_dup_tx_id rows={r[0]}"))(
          one("SELECT SUM(is_dup_tx_id::INT) FROM tx")))
check("tx: 1,997,748 distinct original IDs",
      lambda: (lambda r: (r[0] == 1997748, f"distinct tx_id={r[0]:,}"))(
          one("SELECT COUNT(DISTINCT tx_id) FROM tx")))
check("tx: no generated IDs (all look like TXN + 9 digits)",
      lambda: (lambda r: (r[0] == 0, f"bad ids={r[0]}"))(
          one("SELECT COUNT(*) FROM tx WHERE NOT regexp_full_match(tx_id, 'TXN[0-9]{9}')")))
types = {r[0]: r[1] for r in db.sql("DESCRIBE tx").fetchall()}
check("tx: column types", lambda: (
      types.get("tx_key") == "BIGINT" and types.get("amount_paise") == "BIGINT" and types.get("ts") == "TIMESTAMP",
      f"tx_key={types.get('tx_key')} amount_paise={types.get('amount_paise')} ts={types.get('ts')}"))
check("tx: every src/dst exists in accounts",
      lambda: (lambda r: (r[0] == 0, f"orphans={r[0]}"))(
          one("""SELECT COUNT(*) FROM tx
                 WHERE src NOT IN (SELECT acct_id FROM accounts)
                    OR dst NOT IN (SELECT acct_id FROM accounts)""")))

# ---------- 3. Spot-check 5 rows against the original CSV ----------
def spot():
    rows = db.sql("""
        SELECT t.tx_id, s.acct_no, r.acct_no, t.amount_paise, strftime(t.ts, '%Y-%m-%d %H:%M:%S')
        FROM tx t JOIN accounts s ON t.src = s.acct_id JOIN accounts r ON t.dst = r.acct_id
        USING SAMPLE 5 ROWS""").fetchall()
    mem = duckdb.connect()
    bad = 0
    for tx_id, snd, rcv, paise, ts in rows:
        hit = mem.sql(f"""
            SELECT COUNT(*) FROM read_csv('{CSV}', all_varchar=true)
            WHERE Transaction_ID = '{tx_id}' AND Sender_Account = '{snd}'
              AND Receiver_Account = '{rcv}' AND Timestamp = '{ts}'
              AND round(CAST(Amount AS DOUBLE) * 100) = {paise}""").fetchone()[0]
        bad += (hit != 1)
    mem.close()
    return bad == 0, f"{5 - bad}/5 rows match the CSV exactly"
check("tx: 5 random rows match the CSV", spot)

# ---------- 4. accounts ----------
check("accounts: 24,873 rows, 10 banks",
      lambda: (lambda r: (r == (24873, 10), f"rows={r[0]:,} banks={r[1]}"))(
          one("SELECT COUNT(*), COUNT(DISTINCT bank) FROM accounts")))
check("accounts: all match ^[A-Z]{4}[0-9]{8}$",
      lambda: (lambda r: (r[0] == 0, f"bad={r[0]}"))(
          one("SELECT COUNT(*) FROM accounts WHERE NOT regexp_full_match(acct_no, '[A-Z]{4}[0-9]{8}')")))
check("accounts: no duplicate acct_no",
      lambda: (lambda r: (r[0] == r[1], f"rows={r[0]:,} distinct={r[1]:,}"))(
          one("SELECT COUNT(*), COUNT(DISTINCT acct_no) FROM accounts")))

# ---------- 5. ingest_meta / rejects ----------
check("ingest_meta: latest load 2M loaded, 0 rejected, < 60 s",
      lambda: (lambda r: (r[0] == 2000000 and r[1] == 0 and r[2] < 60,
                          f"loaded={r[0]:,} rejected={r[1]} seconds={round(r[2],1)}"))(
          one("SELECT rows_loaded, rows_rejected, load_seconds FROM ingest_meta ORDER BY loaded_at DESC LIMIT 1")))
check("rejects: empty", lambda: (lambda r: (r[0] == 0, f"rows={r[0]}"))(one("SELECT COUNT(*) FROM rejects")))

# ---------- 6. scores ----------
cols = {r[0] for r in db.sql("DESCRIBE scores").fetchall()}
need = ({f"mp{i}" for i in range(1, 9)} | {f"t{i}" for i in range(1, 8)} |
        {"acct_id","profile_id","mule_index","trust_index","final_index","band","is_flagged",
         "override_applied","l1_score","l2_score","l3_score","victim_score","role","role_confirmed",
         "candidate_roles","upstream_role_share","downstream_role_share","param_points","reasons","ring_id"})
check("scores: all final-spec columns present", lambda: (need <= cols, f"missing: {sorted(need - cols) or 'none'}"))
check("scores: no old m1..m10 columns",
      lambda: (not ({f"m{i}" for i in range(1, 11)} & cols), f"old: {sorted({f'm{i}' for i in range(1,11)} & cols) or 'none'}"))
check("scores: empty", lambda: (lambda r: (r[0] == 0, f"rows={r[0]}"))(one("SELECT COUNT(*) FROM scores")))

# ---------- 7. Constraints ----------
cons = db.sql("""SELECT table_name, constraint_type, constraint_column_names, constraint_text
                 FROM duckdb_constraints()
                 WHERE table_name IN ('tx','scores','layer_links')""").fetchall()
def pk(table):
    return [sorted(c[2]) for c in cons if c[0] == table and c[1] == "PRIMARY KEY"]
check("tx: primary key on tx_key", lambda: (["tx_key"] in pk("tx"), f"PK={pk('tx')}"))
check("layer_links: primary key (profile_id, tx_key)",
      lambda: (["profile_id", "tx_key"] in pk("layer_links"), f"PK={pk('layer_links')}"))
role_check = " ".join(c[3] for c in cons if c[0] == "scores" and c[1] == "CHECK")
check("scores: role CHECK allows VICTIM", lambda: ("VICTIM" in role_check, role_check[:100] or "no CHECK found"))
link_check = " ".join(c[3] for c in cons if c[0] == "layer_links" and c[1] == "CHECK")
check("layer_links: link_type CHECK present",
      lambda: (all(k in link_check for k in ["VICTIM_L1","L1_L2","L2_L2","L2_L3"]), link_check[:100] or "no CHECK found"))

# ---------- 8. scoring_profiles ----------
check("scoring_profiles: one active, locked v1-verified",
      lambda: (lambda r: (r == [("v1-verified", True, True)], f"{r}"))(
          db.sql("SELECT profile_id, is_active, is_locked FROM scoring_profiles").fetchall()))
check("scoring_profiles: definition mentions MP1-MP8 and cash-out categories",
      lambda: (lambda d: (all(k in d for k in ["MP1","MP8","P2A","WALLET_LOAD"]), f"{len(d):,} chars"))(
          one("SELECT definition::VARCHAR FROM scoring_profiles WHERE profile_id='v1-verified'")[0]))

db.close()

# ---------- Report ----------
for status, name, detail in results:
    print(f"[{status:5}] {name:55} {detail}")
print(f"\n{sum(s=='PASS' for s,_,_ in results)} PASS · {sum(s=='FAIL' for s,_,_ in results)} FAIL · "
      f"{sum(s=='ERROR' for s,_,_ in results)} ERROR")
```
